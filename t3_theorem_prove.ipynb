**Задание 1**

Найдите корень уравнения `x**3 + 3*x**2 + 4*x + 2 == 0` с помощью Z3. Сможете вы применить Z3 так, чтобы доказать, что найденный корень единственный, или опровергнуть это утверждение?

```python
# добавимм выражение в 2 разные функций
# зададим ограничения, что для каждой функции должен быть найден корень
# и дополнительно зададим ограничение, что эти корни должны быть разными
# результат unsat, то есть два различных корня не найдено

x = Real('x')

f1 = x**3 + 3*x**2 + 4*x + 2
f2 = x**3 + 3*x**2 + 4*x + 2

solver = Solver()

solver.add(f1 == 0)
solver.add(f2 == 0)
solver.add(f1 != f2)

print(solver.check())
```

**Задание 2**

Доказать правило де Моргана "отрицание дизъюнкции есть конъюнкция отрицаний"  

`p | q == ~ (~p & ~q)`

```python
p = Bool('p')
q = Bool('q')

f = Or(p, q) == Not(And(Not(p), Not(q)))

prove(f)
```

Задача 2. Теорема Пифагора. Доказать `a^2 + b^2 = c^2`

```python
a, b, c = Reals('a b c')

# Уравнение теоремы Пифагора
theorem = a**2 + b**2 == c**2

# Ограничение положительности сторон
restrictions = And(a > 0, b > 0, c > 0)

solver = Solver()
solver.add(theorem, restrictions)

solver.check()
```

Задание 4.

Теорема о трех перпендикулярах. Прямая, проведенная в плоскости через основание наклонной,
перпендикулярна этой наклонной тогда и только тогда, когда она перпендикулярна
проекции наклонной

```python
l1, l2, l3 = Reals('l1 l2 l3')
ah1, ah2, ah3 = Reals('ah1 ah2 ah3')
hb1, hb2, hb3 = Reals('hb1 hb2 hb3')

# прямые, описываемые векторами
l  = [l1, l2, l3]
ah = [ah1, ah2, ah3]
hb = [hb1, hb2, hb3]

# скалярное произведение
def mul(u, v):
    return u[0] * v[0] + u[1] * v[1] + u[2] * v[2]

# AB = AH + HB
ab = [ah[i] + hb[i] for i in range(3)]

s = Solver()

s.add(mul(l, ah) == 0)

# если прямая l перпендикулярна другой, то их произведение = 0
r1  = mul(l, ab) == 0
r2 = mul(l, hb) == 0

s.push()
s.add(r1 != r2) # пытаемся найти контрпример

s.check()
```
